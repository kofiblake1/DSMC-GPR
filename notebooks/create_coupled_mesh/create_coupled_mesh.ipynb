# Create and Visualize Mesh
## Use this section to mesh and visualize 2D object

```python
import gmsh
import numpy as np
import sys
import math

# Initialize gmsh
gmsh.initialize()

title = "wedge_a0a10"
# Create a new model
gmsh.model.add(title)

# Create rectangle points (counterclockwise)
lc = 0.0025  # mesh characteristic length

# Rotation angle in degrees
rotation_angle = 10  # clockwise
rotation_rad = -math.radians(rotation_angle)  # negative for clockwise

# Define base points
base_points = [
    (-1.5e-1, 0),      # p0
    (0, 6e-2),         # p1
    (1.5e-1, 0),       # p2
    (0, -6e-2)         # p3
]

# Rotate points clockwise
rotated_points = []
for x, y in base_points:
    x_rot = x * math.cos(rotation_rad) - y * math.sin(rotation_rad)
    y_rot = x * math.sin(rotation_rad) + y * math.cos(rotation_rad)
    rotated_points.append((x_rot, y_rot))

# Add rotated points to gmsh
points = [gmsh.model.geo.addPoint(x, y, 0, lc) for x, y in rotated_points]

# Create lines connecting the points
lines = [
    gmsh.model.geo.addLine(points[0], points[1]),  # bottom
    gmsh.model.geo.addLine(points[1], points[2]),  # right
    gmsh.model.geo.addLine(points[2], points[3]),  # top
    gmsh.model.geo.addLine(points[3], points[0])   # left
]

# Create curve loop and surface
curve_loop = gmsh.model.geo.addCurveLoop(lines)
surface = gmsh.model.geo.addPlaneSurface([curve_loop])

# Synchronize the model
gmsh.model.geo.synchronize()

# Generate mesh
gmsh.model.mesh.generate(2)

if "-nopopup" not in sys.argv:
    gmsh.fltk.run()

```

# Export Meshes
## This section exports meshes for use in coupled SPARTA/AERO-S simulations. 
The ouptut should be a mesh for SPARTA, a mesh for AERO, and a mapping file that is read by AERO-S that maps the SPARTA elements to AERO-S elements/nodes.

```python
# Get all boundary entities (1D entities forming the boundary)
boundary_entities = gmsh.model.getBoundary([(2, surface)], oriented=False)
print("\nBoundary entities (1D curves):", boundary_entities)

# Extract all 1D elements (lines) on the boundary
dim = 1  # 1D elements
element_types, element_tags, node_tags = gmsh.model.mesh.getElements(dim)

print("\n1D Element Types:", element_types)  # Should be line elements (Type 1 for linear)
print("1D Element Tags:", element_tags)
print("1D Element Node Connectivity:", node_tags)

# Find which elements belong to the boundary
boundary_curves = {e[1] for e in boundary_entities}  # Get boundary curve tags
boundary_elements = []
boundary_node_tags = set()  # Store unique boundary nodes

for etype, etags, enodes in zip(element_types, element_tags, node_tags):
    for i, e in enumerate(etags):
        node_pair = enodes[i * 2: (i + 1) * 2]  # Each line has 2 nodes
        boundary_elements.append((e, node_pair))
        boundary_node_tags.update(node_pair)

# Convert to numpy array for structured access
boundary_node_tags = np.array(list(boundary_node_tags))

# Retrieve coordinates of all nodes
all_node_tags, all_node_coords, _ = gmsh.model.mesh.getNodes()
all_node_coords = np.array(all_node_coords).reshape(-1, 3)

# Find indices of boundary nodes in the full node list
boundary_indices = np.isin(all_node_tags, boundary_node_tags)
boundary_nodes = all_node_coords[boundary_indices]
boundary_tags = boundary_node_tags  # Save original tags

# Create array of 2D coordinates for sorting
coords_2d = boundary_nodes[:, :2]  # Only take x,y coordinates

# Calculate centroid
centroid = np.mean(coords_2d, axis=0)

# Calculate angles from centroid to each point
angles = np.arctan2(coords_2d[:, 1] - centroid[1],
                    coords_2d[:, 0] - centroid[0])

# Sort by angle in clockwise order (note the negative sign)
sorted_indices = np.argsort(-angles)  # Adding negative sign for clockwise order
boundary_nodes = boundary_nodes[sorted_indices]
boundary_node_tags = boundary_tags[sorted_indices]

# After sorting boundary nodes clockwise
sorted_boundary_nodes = boundary_nodes
sorted_boundary_tags = boundary_node_tags

# 1. Save full mesh information (as in meshtest3)
element_types_2d, element_tags_2d, node_tags_2d = gmsh.model.mesh.getElements(2)
topology = []
for etype, etags, enodes in zip(element_types_2d, element_tags_2d, node_tags_2d):
    # For triangles: 3 nodes per element
    if etype == 2:  # 2 = triangle
        num_nodes_per_elem = 3
    elif etype == 3:  # 3 = quadrangle
        num_nodes_per_elem = 4
    else:
        continue
    for i, tag in enumerate(etags):
        nodes = enodes[i * num_nodes_per_elem: (i + 1) * num_nodes_per_elem]
        topology.append((tag, 46, nodes))

mesh_filename = title + "_aero.txt"
with open(mesh_filename, "w") as f:
    # NODE section
    f.write("NODE\n")
    for tag, (x, y, _) in zip(all_node_tags, all_node_coords):
        f.write(f"{tag} {x} {y}\n")
    
    f.write("*\n")
    
    # TOPO section
    f.write("TOPO\n")
    for elem in topology:
        tag, elem_code, nodes = elem
        nodes = list(nodes)
        
        # Get node coordinates
        node_coords = []
        for node in nodes:
            idx = np.where(all_node_tags == node)[0][0]
            node_coords.append(all_node_coords[idx, :2])
        
        # Calculate signed area (positive = counter-clockwise, negative = clockwise)
        signed_area = 0.5 * sum(node_coords[i][0] * node_coords[(i+1) % len(node_coords)][1] - 
                                 node_coords[(i+1) % len(node_coords)][0] * node_coords[i][1] 
                                 for i in range(len(node_coords)))
        
        # If clockwise (negative area), reverse nodes
        if signed_area < 0:
            nodes = list(reversed(nodes))
        
        node_str = " ".join(map(str, nodes))
        f.write(f"{tag} {elem_code} {node_str}\n")
    
    # ATTR section
    f.write("*\n")
    first_elem_tag = topology[0][0]
    last_elem_tag = topology[-1][0]
    f.write("ATTR\n")
    f.write(f"{first_elem_tag} {last_elem_tag} 1\n")
    
    # MATERIAL section
    f.write("*\n")
    f.write("MATERIAL\n")
    f.write("1 0 70e9 0.3 2.7e3 0 3 1.0 0 0 1000 0 0 0 0\n")
    
    # TEMPERATURES section
    f.write("*\n")
    f.write("ITEMPERATURES\n")
    for tag, (x, y, _) in zip(all_node_tags, all_node_coords):
        f.write(f"{tag} {300}\n")

# 2. Save ordered surface representation
output_filename = title + "_sparta.txt"
with open(output_filename, "w") as f:
    num_boundary_nodes = len(sorted_boundary_nodes)
    f.write("Refined Surface (Ordered)\n\n")
    f.write(f"{num_boundary_nodes} points\n")
    f.write(f"{num_boundary_nodes} lines\n\n")
    f.write("Points\n\n")
    
    # Write ordered points
    for i, (x, y, _) in enumerate(sorted_boundary_nodes, 1):
        f.write(f"{i} {x} {y}\n")
    
    f.write("\nLines\n\n")
    # Write lines connecting ordered points
    for i in range(num_boundary_nodes):
        next_i = (i + 1) % num_boundary_nodes
        f.write(f"{i+1} {i+1} {next_i+1}\n")

# 3. Save mapping between ordered indices and original mesh nodes
# Modify the mapping file output section
mapping_filename = title + "_mapping.txt"
with open(mapping_filename, "w") as f:
    f.write("# Ordered_Index Original_Node_Tag\n")
    f.write(f"{len(sorted_boundary_tags)}\n")  # Add total number of nodes
    for i, tag in enumerate(sorted_boundary_tags, 1):
        f.write(f"{i} {tag}\n")

print(f"\n✅ Full mesh data saved to '{mesh_filename}'")
print(f"✅ Ordered surface data saved to '{output_filename}'")
print(f"✅ Boundary node mapping saved to '{mapping_filename}'")

# 4. Save STL representation of the surface
stl_filename = "surface.stl"
with open(stl_filename, "w") as f:
    # STL header
    f.write("solid surface\n")
    
    # Create triangles from boundary nodes
    # We'll create triangles by connecting each boundary edge to the centroid
    centroid_2d = np.mean(sorted_boundary_nodes[:, :2], axis=0)
    centroid = np.array([centroid_2d[0], centroid_2d[1], 0])
    
    for i in range(len(sorted_boundary_nodes)):
        # Get current and next node
        p1 = sorted_boundary_nodes[i]
        p2 = sorted_boundary_nodes[(i + 1) % len(sorted_boundary_nodes)]
        
        # Calculate normal (pointing outward)
        v1 = p2 - p1
        v2 = centroid - p1
        normal = np.cross(v1, v2)
        normal = normal / np.linalg.norm(normal)
        
        # Write facet
        f.write(f"  facet normal {normal[0]:.6e} {normal[1]:.6e} {normal[2]:.6e}\n")
        f.write("    outer loop\n")
        f.write(f"      vertex {p1[0]:.6e} {p1[1]:.6e} {p1[2]:.6e}\n")
        f.write(f"      vertex {p2[0]:.6e} {p2[1]:.6e} {p2[2]:.6e}\n")
        f.write(f"      vertex {centroid[0]:.6e} {centroid[1]:.6e} {centroid[2]:.6e}\n")
        f.write("    endloop\n")
        f.write("  endfacet\n")
    
    # STL footer
    f.write("endsolid surface\n")

print(f"✅ STL surface data saved to '{stl_filename}'")

# Optional: Verify sorting with visualization
import matplotlib.pyplot as plt

# Replace the existing plotting section with:
plt.figure(figsize=(12, 12))

# Add bounding box
box_x = [-2e-1, 2e-1, 2e-1, -2e-1, -2e-1]
box_y = [-2e-1, -2e-1, 2e-1, 2e-1, -2e-1]
plt.plot(box_x, box_y, 'k--', label='Bounding Box')

# Plot boundary nodes and lines
plt.scatter(sorted_boundary_nodes[:, 0], sorted_boundary_nodes[:, 1], c='b')
plt.plot(sorted_boundary_nodes[:, 0], sorted_boundary_nodes[:, 1], 'b-')
# Connect back to first point
plt.plot([sorted_boundary_nodes[-1, 0], sorted_boundary_nodes[0, 0]], 
         [sorted_boundary_nodes[-1, 1], sorted_boundary_nodes[0, 1]], 'b-')

# Add node labels showing both ordered index and original tag
for i, (x, y) in enumerate(sorted_boundary_nodes[:, :2]):
    plt.annotate(f'{i}\n({sorted_boundary_tags[i]})', 
                (x, y),
                xytext=(10, 10),
                textcoords='offset points',
                bbox=dict(facecolor='white', edgecolor='none', alpha=0.7),
                ha='left')

plt.axis('equal')
plt.grid(True)
plt.legend()
plt.title('Boundary Nodes (Clockwise Order)\nFormat: Order_Index(Original_Tag)')
plt.show()

# Separate interior and boundary nodes for plotting
interior_indices = ~boundary_indices
interior_nodes = all_node_coords[interior_indices]

# Print results
print("\nBoundary Elements (Tag and Connectivity):")
for element in boundary_elements:
    print(f"  Element {element[0]}: Nodes {element[1]}")

print("\nBoundary Node Tags:", sorted_boundary_tags)
print("\nBoundary Node Coordinates:")
for tag, coord in zip(sorted_boundary_tags, sorted_boundary_nodes):
    print(f"  Node {tag}: {coord}")

num_boundary_elements = len(boundary_elements)
print(f"\nTotal number of boundary elements: {num_boundary_elements}")

# Cleanup
gmsh.finalize()

# print(f"Number of boundary nodes: {len(boundary_nodes)}")
```
