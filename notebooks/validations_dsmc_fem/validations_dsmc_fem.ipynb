# Vortex Induced Vibration - Cylinder
[1] F.-B. Tian, H. Luo, L. Zhu, J. C. Liao, and X.-Y. Lu, “An efficient immersed boundary-lattice Boltzmann method for the hydrodynamic interaction of elastic filaments,” Journal of Computational Physics, vol. 230, no. 19, pp. 7266–7283, Aug. 2011, doi: 10.1016/j.jcp.2011.05.028.

[2] S. Mittal and S. Singh, “Vortex-induced vibrations at subcritical Re,” J. Fluid Mech., vol. 534, pp. 185–194, Jul. 2005, doi: 10.1017/S0022112005004635.

[3] M. A. Gallis and J. R. Torczynski, “Effect of slip on vortex shedding from a circular cylinder in a gas flow,” Phys. Rev. Fluids, vol. 6, no. 6, p. 063402, Jun. 2021, doi: 10.1103/PhysRevFluids.6.063402.

[4] M. Usami, T. Kamiya, H. Maeda, Y. Furukawa, and T. Abe, “DSMC Calculation of Karman Vortex Flow and Taylor Vortex Flow by an Improved New Collision Scheme (U-system 3),” in AIP Conference Proceedings, Kyoto (Japan): AIP, 2008, pp. 365–370. doi: 10.1063/1.3076503.


```python
import numpy as np

# Constants
kb = 1.380649e-23

# Gas Properties
m = 66.3e-27
n_inf = 3.457e20
T_inf = 273.15
gamma = 5/3
mu = 2.117e-5

# Kinematic Properties
U_inf = 92.37

# Object Dimensions
L = 1


# Calculated Quantities
a = (gamma*kb*T_inf/m)**0.5
M = U_inf/a
print(f"Speed of sound: {a:.2f} m/s")
print(f"Mach number: {M:.2f}")

Re = m*n_inf*U_inf*L/mu
print(f"Reynolds number: {Re:.2e}")

p_inf = n_inf*kb*T_inf
print(f"Pressure: {p_inf:.2f} Pa")

mfp = mu/p_inf*np.sqrt(np.pi*kb*T_inf/(2*m))
print(f"Mean free path: {mfp:.2e} m")
print(f"Knudsen number: {mfp/L:.2e}")

# Computational Resources
grid_size_mfp_ratio = 0.5
grid_size = mfp * grid_size_mfp_ratio

x_min = 0  # m
x_max = 8*L  # m
y_min = 0 # m
y_max = 6*L  # m
lx = x_max - x_min
ly = y_max - y_min
print(f"Domain size: {lx:.2e} m x {ly:.2e} m")

nx = int(lx / grid_size)
ny = int(ly / grid_size)

print(f"Grid size: {grid_size:.2e} m")
print(f"Number of grid points: {nx} x {ny} = {nx*ny}") 

mean_molecular_speed = np.sqrt(8*kb*T_inf/(np.pi*m))
print(f"Mean molecular speed: {mean_molecular_speed:.2f} m/s")

mct = mfp / mean_molecular_speed
print(f"Mean collision time: {mct:.2e} s")

St = 0.15 # Typical for Re < 100
f = St * U_inf / L
print(f"Vortex shedding frequency: {f:.2f} Hz")
print(f"Vortex shedding period: {1/f:.2f} s")

dt = 1e-6
print(f"Time step: {dt:.2e} s")
print(f"Ratio of time step to mean collision time: {dt/mct:.2e}")
print(f"Ratio of time step to vortex shedding period: {dt/(1/f):.2e}")
print(f"Time for 2 oscillations: {2*(1/f):.2f} s")
print(f"Number of time steps for 2 oscillations: {2*(1/f)/dt:.2e}")
```

```python
import numpy as np

Tref = 273.15 #288
pref = 1e-1
kb = 1.380649e-23
n = 3.457e20 #pref/(kb*Tref)
m = 66.3e-27 #4.8e-26
rho = n*m
print(f"Density: {rho:.2e} kg/m^3")
print(f"Number density: {n:.2e} m^-3")

gamma = 5/3 #1.4
a = (gamma*kb*Tref/m)**0.5
Ma = 0.4
v = Ma*a
print(f"Speed of sound: {a:.2f} m/s")
print(f"Flow speed: {v:.2f} m/s")

d_air = 340e-12 #3.8e-10
mfp = 1/(np.sqrt(2)*np.pi*d_air**2*n)
print(f"Mean free path: {mfp:.2e} m")
L = 1

Kn = mfp/L
print(f"Knudsen number: {Kn:.2e}")

Re = np.sqrt(gamma*np.pi/2)*Ma/Kn
print(f"Reynolds number: {Re:.2e}")

xmin = -1
xmax = 5
ymin = -2
ymax = 2

LX = xmax - xmin
LY = ymax - ymin

dx = 0.5*mfp
dy = 0.5*mfp

num_x_cells = LX / dx
num_y_cells = LY / dy
print(f"Number of cells in x direction: {num_x_cells:.2e}")
print(f"Number of cells in y direction: {num_y_cells:.2e}")

num_cells = num_x_cells * num_y_cells
print(f"Total number of cells: {num_cells:.2e}")
```
